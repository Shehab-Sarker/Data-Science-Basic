# Inverse Matrix

```python
d = np.array([
    [0.20, 0.25, 0.75],
    [0.10, 0.02, 0.025],
    [0.20, 0.75, 0.10]
])

```

```python
print(d.shape)
print(np.linalg.inv(d))
```

```python
d = np.array([
    [67,19,21],
    [39,13,14],
    [81,24,26]
])


determine=np.linalg.det(d)
```

```python
print(determine)
```

# eigen value and vector

```python
M=np.array([
    [1,2,3],
    [4,5,6],
    [2,9,5]
])
```

```python
eigvalue,eigvector=np.linalg.eig(M)
print(f"Eigen Value : {eigvalue}")
print(f"Eigen Vector : {eigvector}")
```
