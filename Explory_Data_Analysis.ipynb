# DARA LOAD FOR EDA

```python
import matplotlib.pyplot as plt
import matplotlib.mlab as mlab

import seaborn as sns
import pandas as pd
import numpy as np
```

```python
url="https://raw.githubusercontent.com/FazlyRabbiBD/Data-Science-Book/master/data-dhaka-weather1953-2016.csv"
df=pd.read_csv(url)
df.head()
```

# None Value Chaking

```python
df.isnull().values.any()
```

# Describing statatics

```python
df.describe()
```

# Correlation Function

```python
df.corr()
```

# Using Heatmap

```python
correlation= df.corr()
plt.figure(figsize=(16,8))

sns.heatmap(correlation,annot=True, linewidths=0,
            vmin=-1, cmap="RdBu_r")

plt.show()
```

# Monthly Average Of Rainfall

## by groupby function

```python
rain=df.groupby("Month")["Rainfall"].mean()
rain
```

# Distributino Plot

```python
sns.displot(rain)
```

# Using sorting

```python
rain.sort_values(ascending=False)
```

# Whole datasate Boxplot

```python
plt.figure(figsize=(16,6))

ax=sns.boxplot(data=df,orient="h",palette="Set2")
```

```python
plt.figure(figsize=(16,6))
ax=sns.boxplot(x=df["Rainfall"])
```

```python
df
```

```python
plt.figure(figsize=(16,6))
ax=sns.boxplot(x="RelativeHumidity", y="Rainfall" ,data=df)
```

# Category Boxplot

```python
plt.figure(figsize=(16,6))
ax=sns.catplot(x="Month", y="Rainfall", kind="box",data=df)
```

# Time Series PLOT

```python
df.plot(figsize=(20,10))
plt.title("Time series plot of Dhaka weather")
plt.show()
```

```python
df
```

# Time series of Minimum Temperature change

```python
plt.figure(figsize=(16,6))
plt.title("Time series plot of MinTemp")
ax=sns.lineplot(x="YEAR",y="MinTemp",errorbar=("ci", 10),color="blue",label="MinTemp", data=df)
```

```python
ax
```

# Time series of avg Humidity

```python
rh=df.groupby("YEAR")["RelativeHumidity"].mean()
plt.figure(figsize=(16,6))

ax=sns.lineplot(label="Yearly Mean Relative Humidity", data=rh)
```

```python
plt.figure(figsize=(16,6))
plt.title("Time series of Average Humidity")
ax=sns.lineplot(x="YEAR",y="RelativeHumidity",errorbar=("ci",10),label="RelativeHumidity",data=df)
```
