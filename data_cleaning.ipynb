# Data cleaning

## Loading & Rename

```python
import pandas as pd
import numpy as np
import seaborn as sns

df=pd.read_csv("https://raw.githubusercontent.com/fazlyrabbi77/DataProcessing/master/real-estate.csv")

df
```

## Renaming column with inplace

```python
df.rename(columns={
    "NUM_BEDROOMS": "BEDROOMS",
    "NUM_BATH": "BATH"
}, inplace=True)
df.head()
```

## Check the None Value

```python
df.isnull().values.any()
```

```python
df.isnull().values.sum()
```

## Showing the None values as per attribute

```python
null_column=df.columns[df.isnull().any()]
df[null_column].isnull().sum()
```

```python
print(df[df.isnull().any(axis=1)][null_column].head())
```

## None Value Handiling
### Filling null values with specific value

```python
df.fillna({"PID": 100005000}, inplace=True)

df
```

## Romeve the ROW or Column value

```python
df.drop(["PID"],axis=1,inplace=False)
```

```python
df
```

```python
df.drop([7,8],axis=0,inplace=False)
```

```python
df
```

## Row Wise data Filling

```python
df.loc[2,"ST_NUM"]=197
df.loc[6,"ST_NUM"]=208
df
```

## Unwanted Value treatment

```python
cnt=0
for row in df["OWN_OCCUPIED"]:
    try:
        int(row)
        df.loc[cnt, "OWN_OCCUPIED"]=np.nan
    except ValueError:
        pass
    cnt+=1


df
```

```python
cnt=0
for row in df["BATH"]:
    try:
        float(row)
        
    except ValueError:
        df.loc[cnt,"BATH"]=np.nan
    
    cnt+=1
    
df
```

```python
df["BEDROOMS"]=pd.to_numeric(df["BEDROOMS"],errors="coerce")
df["BATH"]=pd.to_numeric(df["BATH"],errors="coerce")
df["SQ_FT"]=pd.to_numeric(df["SQ_FT"],errors="coerce")
df
```

## fiill the None value By MODE Value

```python
df["OWN_OCCUPIED"] = df["OWN_OCCUPIED"].fillna(
    df["OWN_OCCUPIED"].mode()[0]
)
```

```python
df
```

## Use of Median
### Group by permeter check

```python
df.groupby("BEDROOMS")["SQ_FT"].median()
```

## Filling Null with group by VParameter

```python
df["SQ_FT"]=df["SQ_FT"].fillna(df.groupby("BEDROOMS")["SQ_FT"].transform("median"))
df
```

```python
df["SQ_FT"]=df["SQ_FT"].fillna(df["SQ_FT"].median())
```

```python
df
```

```python
df.loc[8,"BEDROOMS"]=3
df.loc[2,"BEDROOMS"]=1
df.loc[5,"BEDROOMS"]=1
```

```python
df
```

```python
df["BATH"]=df["BATH"].bfill()
df
```

## converting street number to  int

```python
df.PID=df.PID.astype("int64")
df.ST_NUM=df.ST_NUM.astype("int64")
df.BEDROOMS=df.BEDROOMS.astype("int64")
df.BATH=df.BATH.astype("int64")
df.SQ_FT=df.SQ_FT.astype("int64")
```
