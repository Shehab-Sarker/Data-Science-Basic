```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x=np.arange(11)
y=x**2 
print(x,y)
```

```python
plt.figure(figsize=(8,6))
plt.scatter(x,y)

plt.plot(x,y)
plt.xlabel("X")
plt.ylabel("Y")

plt.title("x vs y")

plt.show()
```

```python
plt.scatter(x,y)
```

```python
plt.plot(x,y ,"r-*")
```

```python
plt.figure(figsize=(8,6))
plt.scatter(x,y)

plt.plot(x,y)
plt.xlabel("X")
plt.ylabel("Y")

plt.title("x vs y")

plt.xticks(range(0,8))
plt.yticks(range(0,100,10))

plt.show()
```

```python
plt.subplot( 1,3,1)
plt.plot(x,y,"r-*") 

plt.subplot( 1,3,2)
plt.plot(y,x,"b-*")

plt.subplot( 1,3,3)
plt.plot(x,y,"g--")
```

```python
plt.subplot( 2,3,1)
plt.plot(x,y,"r-*") 

plt.subplot( 2,3,2)
plt.plot(y,x,"b-*")

plt.subplot( 1,3,3)
plt.plot(x,y,"g--")

plt.subplot( 2,3,4)
plt.plot(x,y,"r-*") 

```

```python
fig=plt.figure(figsize=(10,5))

axes=fig.add_axes([0,0,1,1])

axes.plot(x,y, "r-*")
axes.set_xlabel("X_label")
axes.set_ylabel("Y_label")

axes.set_title("Save Test Fugure")

fig.savefig("graph.pdf", dpi=200)
```

```python
fig=plt.figure(figsize=(10,5))

axes= fig.add_axes([0,0,1,1])

axes.plot(y,x,"b-^")
axes.set_xlabel("X_label")
axes.set_ylabel("Y_label")

axes.set_title("Test Figure")

fig.savefig("graph2.pdf",dpi=200)

plt.show
```
